Imports

```python
import matplotlib.pyplot as plt
from src.data_loader import DataLoader
from src.signals import MomentumSignals
loader = DataLoader()
signal_generator = MomentumSignals()
```

Defining tickers for quick editing

```python
TICKERS = [
    "AAPL",
    "MSFT",
    "AMZN",
    "NVDA",
    "GLD"
]
```

Downloading data and a quick preview

```python
prices = loader.download_prices(TICKERS)

prices.head()
```

Plotting the closing prices

```python
prices.plot(
    figsize=(12, 6),
    title="Asset Prices"
)

plt.show()
```

Computing the returns on the assets

```python
returns = loader.compute_returns(prices)

returns.head()
```

Computing the momentum signals

```python
signals = signal_generator.compute_momentum_signal(
    prices,
    lookback=252
)

signals.tail()
```

Plotting the momentum signal

```python
signals["AAPL"].plot(
    figsize=(12, 4),
    title="AAPL Momentum Signal"
)

plt.show()
```
