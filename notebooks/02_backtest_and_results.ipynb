Imports

```python
import pandas as pd
import matplotlib.pyplot as plt
from src.data_loader import DataLoader
from src.signals import MomentumSignals
from src.backtest import Backtester
from src.performance import PerformanceMetrics

loader = DataLoader()
signal_generator = MomentumSignals()
backtester = Backtester()
perf_calculator = PerformanceMetrics()
```

Defining tickers for quick editing, adding SPY for benchmarking later

```python
TICKERS = [
    "AAPL",
    "MSFT",
    "AMZN",
    "NVDA",
    "GLD"
]
```

Downloading data and a quick preview

```python
prices = loader.download_prices(TICKERS)

prices.head()
```

Computing the returns on the assets

```python
returns = loader.compute_returns(prices)

returns.head()
```

Computing the momentum signals

```python
signals = signal_generator.compute_momentum_signal(
    prices,
    lookback=252
)

signals.tail()
```

Computing the strategy returns for each asset

```python
strategy_returns = backtester.compute_strategy_returns(
    returns,
    signals
)

strategy_returns.head()
```

Aggregating the return across the portfolio

```python
portfolio_returns = backtester.aggregate_portfolio_returns(
    strategy_returns
)

portfolio_returns.head()
```

Computing the cumulative returns

```python
cumulative_returns = backtester.compute_cumulative_returns(
    portfolio_returns
)

cumulative_returns.tail()
```

Plotting the Time-Series Momentum strategy cumulative returns

```python
plt.figure(figsize=(12, 6))

plt.plot(
    cumulative_returns,
    label="TSMOM"
)

plt.title(
    "Cumulative returns of TSMOM strategy"
)

plt.xlabel("Date")
plt.ylabel("Cumulative Returns")

plt.legend()

plt.grid(True)

plt.show()
```

Adding SPY as benchmark for the strategy and aligning its starting period with the portfolio

```python
spy_prices = loader.download_prices(["SPY"])

spy_returns = loader.compute_returns(spy_prices)

benchmark_returns = spy_returns["SPY"]

benchmark_start = benchmark_returns.index.min() + pd.DateOffset(years=1)

benchmark_returns = benchmark_returns.loc[
    benchmark_returns.index >= benchmark_start
]

benchmark_cumulative = (
    1 + benchmark_returns
).cumprod()

aligned_strategy = cumulative_returns.loc[
    cumulative_returns.index >= benchmark_start
]

plt.figure(figsize=(12, 6))

plt.plot(
    aligned_strategy,
    label="TSMOM Portfolio"
)

plt.plot(
    benchmark_cumulative,
    label="SPY"
)

plt.title(
    "Strategy vs SPY"
)

plt.xlabel("Date")
plt.ylabel("Cumulative Returns")

plt.legend()

plt.grid(True)

plt.show()
```

Calculating performance metrics

```python
ann_return = perf_calculator.annualized_return(
    portfolio_returns
)

ann_vol = perf_calculator.annualized_volatility(
    portfolio_returns
)

sr = perf_calculator.sharpe_ratio(
    portfolio_returns
)

mdd = perf_calculator.max_drawdown(
    cumulative_returns
)

print("PERFORMANCE:")
print(f"Annualized Return: {ann_return:.2%}")
print(f"Annualized Volatility: {ann_vol:.2%}")
print(f"Sharpe Ratio: {sr:.2f}")
print(f"Maximum Drawdown: {mdd:.2%}")
```
